# 03. Class Imbalance Handling
SMOTE, Borderline-SMOTE, ADASYN, and cost-sensitive sampling experiments.

```python
import scipy.sparse
import joblib
import numpy as np
import pandas as pd
from imblearn.over_sampling import BorderlineSMOTE

# Load saved data
X_train = scipy.sparse.load_npz("../data/processed/X_train_selected.npz")
X_test  = scipy.sparse.load_npz("../data/processed/X_test_selected.npz")
y_train = np.load("../data/processed/y_train.npy")
y_test  = np.load("../data/processed/y_test.npy")

preprocessor = joblib.load("../data/processed/preprocessor.joblib")
selector    = joblib.load("../data/processed/selector.joblib")

# Initialise Borderline-SMOTE
smote = BorderlineSMOTE(
    sampling_strategy='auto',
    random_state=42,
    kind='borderline-1'
)

# Resample using X_train (not X_train_processed)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

# Sanity check
print("Before SMOTE – class distribution:")
print(pd.Series(y_train).value_counts())

print("\nAfter SMOTE – class distribution:")
print(pd.Series(y_train_balanced).value_counts())
```

```python
import os
import scipy.sparse
import numpy as np

output_dir = "../data/processed"

# Save the resampled arrays already in memory
scipy.sparse.save_npz(
    os.path.join(output_dir, "X_train_resampled.npz"),
    X_train_balanced
)
np.save(
    os.path.join(output_dir, "y_train_resampled.npy"),
    y_train_balanced
)

print("Saved successfully! No need to rerun SMOTE.")
```
